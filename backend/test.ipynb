### **En cas de danger immédiat**

### **Préparation d'un sac d'urgence**
- Préparez un sac avec des affaires essentielles pour vous et vos enfants (vêtements, médicaments, clés, copies de papiers, un peu d'argent).<br>- Conservez-le dans un endroit sûr, idéalement chez une personne de confiance, si sa découverte pourrait vous mettre en danger.<br>- **Ne retardez pas votre mise à l'abri** pour préparer ou récupérer ce sac.<br><br>
### **Documents importants à rassembler**
- **Identité et famille** : pièces d'identité, livret de famille, permis de séjour.<br>- **Santé et assurances** : cartes d'assurance maladie, certificats d'assurance.<br>- **Travail et finances** : cartes bancaires, contrats de travail, fiches de salaire, extraits de comptes, justificatifs de charges, dettes, documents immobiliers, déclarations d'impôts.<br>- **Preuves des violences** : constats médicaux, lettres, e-mails, messages attestant des faits et de leurs conséquences psychiques.<br>- **Conservation** : Gardez les copies dans un lieu sûr, hors de portée de votre partenaire.
